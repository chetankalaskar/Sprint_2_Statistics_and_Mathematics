# Probability Distributions


### What is a Probability Distribution?
A Probability Distribution describes how the possible values of a random variable are distributed and the probability of each value occurring.

It helps us understand the likelihood of different outcomes in a random experiment.

### Bernoulli Distribution
A Bernoulli Distribution models an experiment with only two possible outcomes:

* Success (1)
* Failure (0)

Example:

A customer visits the online shopping website.

* Purchased the product → 1 (Success)
* Did not purchase the product → 0 (Failure)

```python
import numpy as np

purchase = np.random.binomial(n=1, p=0.7)

print("Purchase:", purchase)
```

Each customer can have only one of two outcomes: purchase or no purchase.

### Binomial Distribution

A Binomial Distribution models the number of successes in a fixed number of independent Bernoulli trials.

Example:

Suppose 10 customers visit the website.

Each customer has a 70% chance of purchasing a product.

We want to know how many customers make a purchase.


```python
import numpy as np

purchases = np.random.binomial(n=10, p=0.7)
print("Customers who purchased:", purchases)
```

The output shows the number of successful purchases among 10 customers.

### Poisson Distribution
A Poisson Distribution models the number of events occurring in a fixed interval of time or space.

Example:

The website receives an average of 5 customer orders per hour.

We want to know how many orders arrive in the next hour.

```python
import numpy as np

orders = np.random.poisson(lam=5)
print("Orders received:", orders)
```

The output represents the number of orders received in one hour.

### Uniform Distribution
A Uniform Distribution means every value within a given range has an equal probability of occurring.

Example:

The website gives a random discount between 10% and 30% to customers.

Every discount value has an equal chance of being selected.

```python
import numpy as np

discount = np.random.uniform(10, 30)
print("Discount:", round(discount,2), "%")
```

Every discount between 10% and 30% is equally likely.

### Normal Distribution
A Normal Distribution is a bell-shaped distribution where most observations are close to the average.

Example:

The purchase amount of customers is approximately normally distributed.

* Average purchase = 100
* Most customers spend around 100
* Very few customers spend extremely low or extremely high amounts.

```python
import numpy as np

purchase_amount = np.random.normal(loc=100, scale=20)
print("Purchase Amount:", round(purchase_amount,2))
```

Most purchase amounts are close to the average value.

### Exponential Distribution
An Exponential Distribution models the waiting time between two consecutive events.

Example:

The website measures the time until the next customer places an order.

```python
import numpy as np

waiting_time = np.random.exponential(scale=5)
print("Waiting Time:", round(waiting_time,2), "minutes")
```

The output represents the expected waiting time until the next customer order.
