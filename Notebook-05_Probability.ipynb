# Probability

### What is Probability?
Probability is the measure of how likely an event is to happen.

The value of probability always lies between 0 and 1.

* 0 → Impossible event
* 1 → Certain event

Formula:

P(E)= Number of Favorable Outcomes / Total Number of Outcomes

Example:

Consider rolling a fair six-sided dice.

Possible outcomes: {1, 2, 3, 4, 5, 6}

Probability of getting 4 : P(4)=1/6


```python
import random

print("Dice Roll:", random.randint(1,6))
```

### Events
An Event is an outcome or a set of outcomes of an experiment.

Example:

Rolling a dice

Event A:

Getting an even number : {2,4,6}

Probability : 3/6 = 0.5

```python
event = [2,4,6]
print(event)
```

### Sample Space
The Sample Space is the set of all possible outcomes of an experiment.

Example:

Rolling one dice

S = {1,2,3,4,5,6}

```python
sample_space = [1,2,3,4,5,6]

print(sample_space)
```

### Independent Events
Two events are Independent if the occurrence of one does not affect the occurrence of the other.

Example:

Roll a dice and toss a coin.

The dice result does not affect the coin result.

```python
import random

dice = random.randint(1,6)
coin = random.choice(["Head","Tail"])

print(dice)
print(coin)
```

### Dependent Events
Two events are Dependent if the occurrence of one event affects the probability of the other.

Example:

A bag contain,

* 3 Red Balls
* 2 Blue Balls

If one red ball is removed without replacement, the probability of drawing another red ball changes.

```python
balls = ["R","R","R","B","B"]

first = balls.pop(0)

print("Remaining:", balls)
```

### Conditional Probability
Conditional Probability is the probability of an event occurring given that another event has already occurred.

Formula: 

P(A∣B) = P(A∩B) / P(B)

Example:

A dice is rolled.

Given that the number is even, what is the probability that it is greater than 4?

Even numbers : {2,4,6}

Greater than 4 : {6}

Probability : 1/3

```python
even = [2,4,6]

greater_than_four = [x for x in even if x > 4]

print(len(greater_than_four)/len(even))
```

### Bayes' Theorem
Bayes' Theorem calculates the probability of an event using prior knowledge.

Formula:
P(A∣B) = (P(B∣A)×P(A)) / P(B)

Example:

Suppose

P(Disease)=0.01

P(Positive|Disease)=0.99

P(Positive)=0.05

Then

P(Disease|Positive)

=(0.99×0.01)/0.05

=0.198

```python
P_A = 0.01
P_B_given_A = 0.99
P_B = 0.05

bayes = (P_B_given_A * P_A) / P_B

print(bayes)
```

### Probability Rules
Probability follows some basic rules.

Rule 1:

0 ≤ P(E) ≤ 1

Rule 2:

P(Sample Space)=1

Rule 3:

P(Not A)=1-P(A)

Example:

Probability of getting an even number

3/6 = 0.5

Probability of not getting an even number

1−0.5=0.5

```python
p_even = 3/6

print("p_odd :",1-p_even)
```

### Random Variables

Example:

Possible values : 1,2,3,4,5,6

```python
import random

X = random.randint(1,6)

print(X)
```

### Expected Value
Expected Value is the average outcome expected after repeating an experiment many times.

Formula:

E(X)=∑xP(x)

Example:

Fair Dice: (1+2+3+4+5+6)/6 = 3.5

Expected Value : 3.5

```python
outcomes = [1,2,3,4,5,6]

expected = sum(outcomes)/len(outcomes)

print(expected)
```
