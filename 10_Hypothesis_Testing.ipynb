# Hypothesis Testing

### What is Hypothesis Testing?
Hypothesis Testing is a statistical method used to determine whether there is enough evidence to accept or reject a claim about a population.


It is commonly used to test claims, differences, relationships, or effects.

### Basic Idea

We start with a claim called the Null Hypothesis (H₀) and use sample data to determine whether there is enough evidence to reject it.


Example

A company claims that its battery has an average life of 10 hours.

We collect battery-life measurements from a sample: 

```python
# sample dataset

sample = [9.5, 10.2, 9.8, 10.1, 9.7, 10.3, 9.9, 10.0, 9.6, 10.1]
```

We use this sample to investigate whether the company's claim is supported by the data.

### Null Hypothesis (H₀)
The Null Hypothesis (H₀) is the default assumption that there is no significant difference or effect.

It commonly states that:

- There is no significant difference.
- There is no significant effect.
- There is no relationship.
- A population parameter equals a specified value.


### Important Point

The null hypothesis is the statement we test using statistical evidence.

Example:

The company claims:

Average battery life = 10 hours

So,

H₀ : μ = 10 hours

Where:
- H₀ = Null Hypothesis
- μ = Population mean
- 10 = Claimed population mean

```python
claim = 10

print("Null Hypothesis: Average battery life =", claim, "hours")
```

### Alternative Hypothesis (H₁)
The Alternative Hypothesis (H₁) states that there is a significant difference, effect or relationship.

It is considered when the evidence is strong enough to reject the null hypothesis

### Important Point

The alternative hypothesis determines the direction of the test.

Example:

Suppose we believe the average battery life is not equal to 10 hours.

H₁ : μ ≠ 10 hours

```python
print("Alternative Hypothesis: Average battery life is not 10 hours")
```

### p-value
The p-value is the probability of obtaining the observed result if the null hypothesis is true.

A small p-value indicates that the observed result would be relatively unusual under H₀.

Decision:

* p-value ≤ 0.05 → Reject H₀
* p-value > 0.05 → Fail to reject H₀



### Important Point

A p-value is not the probability that the null hypothesis is true.

It is calculated under the assumption that the null hypothesis is true.


Example:

Suppose, p-value = 0.03

Since, 0.03 < 0.05

Reject the null hypothesis.

```python
p_value = 0.03

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

### Confidence Interval
A Confidence Interval (CI) is a range of values that is likely to contain the true population parameter based on sample data.

A 95% confidence interval means we are 95% confident that the true population mean lies within the calculated interval.

Example:

Battery life confidence interval:

9.7 hours to 10.2 hours

This provides a range of values that are compatible with the data under the statistical model used.

```python
import numpy as np

sample = [9.5, 10.2, 9.8, 10.1, 9.7, 10.3, 9.9, 10.0, 9.6, 10.1]

print("Mean:", np.mean(sample))
```

### Significance Level (α)
The significance level (α) is the threshold used to decide whether the evidence against the null hypothesis is statistically significant.

The most commonly used value is:

α = 0.05

Example:

If α = 0.05

There is a 5% chance of rejecting a true null hypothesis.

```python
alpha = 0.05

print("Significance Level:", alpha)
```

### Interpretation

When α = 0.05, the test is designed to limit the long-run probability of a Type I Error to 5%, under the assumptions of the test.

### Type I Error
A Type I Error occurs when we reject the null hypothesis even though it is actually true.

It is also called a False Positive.

Example:

The battery actually lasts 10 hours, but we incorrectly conclude that it does not.

```python
print("Type I Error: Rejecting a true Null Hypothesis")
```

### Type II Error
A Type II Error occurs when we fail to reject the null hypothesis even though it is false.

It is also called a False Negative.

Example:

The battery actually lasts 9 hours, but we incorrectly conclude that it still lasts 10 hours.

```python
print("Type II Error: Failing to reject a false Null Hypothesis")
```

### One-tailed Test
A One-tailed Test checks whether a parameter is greater than or less than a specific value.

A one-tailed test checks for an effect in one specific direction.

There are two types:

- Right-tailed test: A right-tailed test checks whether the population parameter is greater than a specified value.
- Left-tailed test: A left-tailed test checks whether the population parameter is less than a specified value.

Right-tailed Test, H₁ : μ > 10

Left-tailed Test, H₁ : μ < 10

Example:

A company claims the battery lasts more than 10 hours.

```python
sample_mean = 10.2

if sample_mean > 10:
    print("Supports Right-Tailed Test")
```

### Two-tailed Test
A Two-tailed Test checks whether a parameter is different from a specified value in either direction.

H₁ : μ ≠ 10


## Important Point

Again, a sample mean being different from 10 does not automatically mean the difference is statistically significant. The hypothesis test and p-value determine the statistical evidence.


Example:

We simply want to know whether the battery life is different from 10 hours, regardless of whether it is higher or lower.

```python
sample_mean = 9.7

if sample_mean != 10:
    print("Supports Two-Tailed Test")
```
