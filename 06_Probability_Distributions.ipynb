# Probability Distributions


### What is a Probability Distribution?
A Probability Distribution describes how the possible values of a random variable are distributed and the probability of each value occurring.

It helps us understand the likelihood of different outcomes in a random experiment.


Probability distributions are important in Data Science, Artificial Intelligence, and Machine Learning because real-world data contains uncertainty.

They help us:

- Understand data
- Model uncertainty
- Generate simulated data
- Detect unusual observations
- Estimate probabilities
- Make predictions
- Perform statistical tests
- Understand errors and variation
- Build statistical and machine learning models


### Bernoulli Distribution
A Bernoulli Distribution models an experiment with only two possible outcomes:



Example:
- Customer purchases a product or not.
- Student passes or fails.
- Email is spam or not spam.
- Machine works or fails.
- Loan is approved or rejected.
- Patient has a disease or does not have it.

### Key Characteristics
* It has only two possible outcomes.
* The outcomes are usually represented as 1 and 0.
* Probability of success = p
* Probability of failure = 1 - p
* There is only one trial.

### Parameters
- n	Number of trials, which is 1 for Bernoulli
- p	Probability of success


### What Happens When It Goes Wrong?

If the probability p is incorrectly estimated, the model may give incorrect expectations.

For example, suppose the actual purchase probability is 40%, but we assume it is 70%.

This can cause:

- Incorrect sales predictions
- Incorrect inventory planning
- Poor marketing decisions



Example:

A customer visits the online shopping website.
* Success (1)
* Failure (0)


```python
import numpy as np

purchase = np.random.binomial(n=1, p=0.7)

print("Purchase:", purchase)
```

Each customer can have only one of two outcomes: purchase or no purchase.

### Binomial Distribution

A Binomial Distribution models the number of successes in a fixed number of independent Bernoulli trials.

Used when we want to count how many times an event is successful.



### Key Characteristics

Binomial Distribution extends the idea of Bernoulli Distribution.

It counts the number of successes in multiple independent trials.

For example:

10 customers

     ↓
Each customer either purchases or does not purchase

     ↓
Count total purchases




### Conditions for Binomial Distribution

There are four important conditions:

- Fixed number of trials
- Each trial has only two outcomes
- Trials are independent
- Probability of success remains constant


### Parameters
- n	Number of trials
- p	Probability of success
- 1 - p	Probability of failure


Examples:

Number of customers who purchase
Number of students who pass
Number of defective products
Number of successful calls
Number of patients responding to treatment


### Real-Life Applications


### Quality Control

Suppose a factory produces 100 products.

If the probability that a product is defective is 2%, we can estimate how many defective products may occur among 100 products.

### Marketing

Suppose an advertisement has a 10% chance of generating a sale.

For 1,000 customers, Binomial Distribution can model the number of successful sales.

### What Happens When It Goes Wrong?

Binomial Distribution should not be used blindly.

If the trials are not independent, the results may be misleading.

For example, suppose 10 customers are from the same family and their purchasing decisions influence each other.

The assumption of independent trials may not hold.

This can result in:

- Incorrect probability calculations
- Incorrect business predictions
- Incorrect risk estimates



### Bernoulli vs Binomial

Bernoulli
- One trial
- Two possible outcomes
- Example: One customer purchases  
- n = 1 


Binomial     
- Multiple trials
- Counts successes
- Example: 100 customers, count purchases
- n > 1



### Important Point

Bernoulli asks "Did it happen?"

Binomial asks "How many times did it happen?"


Example:

Suppose 10 customers visit the website.

Each customer has a 70% chance of purchasing a product.

We want to know how many customers make a purchase.

```python
import numpy as np

purchases = np.random.binomial(n=10, p=0.7)
print("Customers who purchased:", purchases)
```

The output shows the number of successful purchases among 10 customers.

### Poisson Distribution
A Poisson Distribution models the number of events occurring in a fixed interval of time or space.

Used for counting events.

The interval can be:
- Time
- Distance
- Area
- Volume

Examples:

- Number of orders per hour
- Number of calls per minute
- Number of accidents per month
- Number of website visits per minute


### What Happens When It Goes Wrong?

Poisson Distribution assumes that the event rate is reasonably stable over the interval being modeled.

May cause:

- Incorrect server capacity planning
- Incorrect staffing
- Poor resource allocation

### Important Point

Poisson Distribution answers: "How many events occur in a fixed interval?"

Example:

The website receives an average of 5 customer orders per hour.

We want to know how many orders arrive in the next hour.

```python
import numpy as np

orders = np.random.poisson(lam=5)
print("Orders received:", orders)
```

The output represents the number of orders received in one hour.

### Uniform Distribution
A Uniform Distribution means every value within a given range has an equal probability of occurring.

There are two common types:
- Discrete Uniform Distribution
- Continuous Uniform Distribution


### Discrete Uniform Distribution
- Deals with countable values.
- Each possible value has equal probability.
- Individual values can have non-zero probability.
- Example: Rolling a fair die.
- Formula: P(X = x) = 1/n.


### Continuous Uniform Distribution
- Deals with continuous values within an interval.
- All values have equal probability density.
- An individual exact value has zero probability.
- Example: Selecting a random number between 0 and 1.
- Formula: f(x) = 1/(b-a).



### Parameters

For the example:

- Lower limit = 10
- Upper limit = 30


Uses

Uniform Distribution is useful when we have no reason to believe that one value is more likely than another within a range.

Examples:

- Random number generation
- Simulation
- Random sampling
- Testing algorithms
- Computer simulations


### Real-Life Applications
- Simulation

Suppose delivery time can be anywhere between 20 and 40 minutes and we assume every value within that range is equally likely.

We can use a Uniform Distribution to simulate delivery times.


- Computer Science

Random values are frequently generated for:

- Games
- Simulations
- Testing
- Algorithms


### What Happens When It Goes Wrong?

A real-world process may not actually have equal probability density across the range.

For example, suppose we assume customer discounts between 10% and 30% are equally likely.

But the actual company system gives:

- 10% → 50% of customers
- 20% → 30% of customers
- 30% → 20% of customers

Then Uniform Distribution would not accurately represent the actual data.

This can lead to:

- Incorrect simulations
- Incorrect probability estimates
- Poor predictions
- Important Point

Uniform Distribution is useful when values within a range are treated as equally likely.


Example:

The website gives a random discount between 10% and 30% to customers.

Every discount value has an equal chance of being selected.

```python
import numpy as np

discount = np.random.uniform(10, 30)
print("Discount:", round(discount,2), "%")
```

Every discount between 10% and 30% is equally likely.

### Normal Distribution
A Normal Distribution is a bell-shaped distribution where most observations are close to the average.


### Key Characteristics

A Normal Distribution has a bell-shaped curve.


Normal Distribution is widely used for:

- Statistical analysis
- Measurement data
- Error analysis
- Quality control
- Hypothesis testing
- Confidence intervals
- Machine learning


### Real-Life Applications
- Student Scores

If exam scores are approximately normally distributed, many students may score around the average, while very high and very low scores are less common.

- Manufacturing

A machine produces components with a target length of 10 cm.

Small manufacturing variations may approximately follow a normal distribution.

- Data Science

Normal Distribution is commonly used in statistical modeling and can be useful for understanding variation and measurement errors.

### What Happens When It Goes Wrong?

Not every dataset follows a normal distribution.

Some datasets may be:

- Skewed
- Heavy-tailed
- Multimodal
- Contain many outliers

For example, income data is often strongly right-skewed rather than normally distributed.

If we incorrectly assume normality, it can lead to:

- Incorrect statistical conclusions
- Incorrect confidence intervals
- Incorrect hypothesis tests
- Poor modeling assumptions


### Important Point

Normal Distribution is useful when data tends to cluster around a central value with fewer observations toward the extremes.

Important characteristics:

- It is symmetrical around the mean.
- Mean = Median = Mode.
- Most observations are near the mean.
- Fewer observations occur far from the mean.
- The shape is determined by the mean and standard deviation.


Example:

The purchase amount of customers is approximately normally distributed.

* Average purchase = 100
* Most customers spend around 100
* Very few customers spend extremely low or extremely high amounts.

```python
import numpy as np

purchase_amount = np.random.normal(loc=100, scale=20)
print("Purchase Amount:", round(purchase_amount,2))
```

Most purchase amounts are close to the average value.

### Exponential Distribution
An Exponential Distribution models the waiting time between two consecutive events.


It is closely related to the Poisson Distribution.

Think of them as:

Poisson       → How many events?

Exponential   → How long until the next event?


### Key Characteristics

Exponential Distribution is commonly used to model:

- Waiting time
- Time between events
- Time until an event occurs

It is usually right-skewed, meaning shorter waiting times are more common while very long waiting times are possible.

Exponential Distribution is commonly used for:

- Waiting times
- Customer arrivals
- Machine failures
- Service times
- Network requests
- Reliability analysis


### Real-Life Applications
- Bank

A bank may want to model the time between customers arriving at a particular counter.

- Machine Maintenance

Suppose machines fail randomly over time.

Exponential Distribution can be used in certain reliability models to represent the waiting time until failure.

- Website

If website requests arrive at an approximately constant average rate, the time between requests can sometimes be modeled using an Exponential Distribution.

### What Happens When It Goes Wrong?

Exponential Distribution assumes a particular type of event-arrival process, commonly associated with a constant event rate.

If the arrival rate changes significantly over time, the model may not accurately represent waiting times.

For example:

- 9 AM  → Many customers
- 2 PM  → Few customers
- 6 PM  → Many customers

Using one constant rate may produce inaccurate results.

This can lead to:

- Incorrect waiting-time predictions
- Incorrect staffing decisions
- Poor resource planning
- Incorrect reliability estimates


### Important Point

Exponential Distribution answers: "How long do we wait until the next event?"


Example:

The website measures the time until the next customer places an order.

```python
import numpy as np

waiting_time = np.random.exponential(scale=5)
print("Waiting Time:", round(waiting_time,2), "minutes")
```

The output represents the expected waiting time until the next customer order.
