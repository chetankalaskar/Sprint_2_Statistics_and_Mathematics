# 01_Introduction_to_Statistics

### What is Statistics?
* Statistics is the branch of mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data to make better decisions.
* Simply put, statistics helps us understand data and find useful information from it.

Example:

A hospital wants to know the average age of patients visiting every day. It collects patient ages, calculates the average, and uses the result to plan healthcare services.

### Why Statistics is Important in AI/ML
Statistics is the foundation of Artificial Intelligence and Machine Learning because AI models learn patterns from data. Statistics helps understand data before building models.

Importance:

* Understand data distribution
* Handle missing values
* Detect outliers
* Select important features
* Evaluate model performance
* Make predictions

Example:

A bank wants to predict whether a customer will repay a loan.

Statistics helps to:

* Analyze customer income
* Calculate average salary
* Find risky customers
* Understand relationships between variables

Machine Learning then uses this information to predict loan approval.

### Statistics vs Machine Learning

* Statistics focuses on understanding data meanwhile Machine Learning focuses on making predictions.
* Statistics explains relationships but Machine Learning learns patterns automatically
* Statistics uses mathematical models while  Machine Learning uses algorithms
* Statistics smaller datasets are sufficient  meanwhile Machine Learning usually requires larger datasets

Example:

Statistics:
A teacher calculates the average marks of students.

Machine Learning:
A system predicts whether a student will pass based on previous records.

### Descriptive Statistics
Descriptive statistics summarizes and describes data using numbers and graphs.

Common measures:

* Mean
* Median
* Mode
* Minimum
* Maximum
* Standard Deviation

Example:
A company calculates:

* Average employee salary
* Highest salary
* Lowest salary

to understand employee pay


```python
import pandas as pd

marks = pd.Series([40, 25, 45, 50, 44])

print(marks.describe())
```

### Inferential Statistics
Inferential statistics uses a sample to make conclusions about an entire population.

Instead of studying everyone, we study a small group.

Example:

A company surveys 500 customers and predicts what all customers think about its product.

### Population
Population is the complete group of people, objects, or observations we want to study.

Example:

All employees in a company.

All students in a college.


### Sample
A sample is a smaller group selected from the population for analysis.

Example:

Instead of surveying all 20,000 customers, a company surveys 500 customers.

### Census
A census collects information from every member of the population.

Example:

The Indian Census collects information from every household in the country.

### Sampling
Sampling is the process of selecting a subset of the population for study.

Example:

A food company tastes products using 100 customers before launching them nationwide

```python
import pandas as pd

data = pd.DataFrame({
    "Age":[11,12,13,14,15,16,17,18]
})

sample = data.sample(3)

print(sample)
```

### Features
Features are input variables used by a Machine Learning model to make predictions.
Features are also called Independent Variables.

Example:

Predicting house price.

Features:

* Area
* Number of Bedrooms
* Location

Target:

House Price



```python
import pandas as pd

data = pd.DataFrame({
    "Area":[5000,600,750],
    "Bedrooms":[2,3,4],
    "Price":[25,30,40]
})

X = data[["Area","Bedrooms"]]
y = data["Price"]

print(X)
```

### Observation
An observation is a single row in a dataset.

Each observation represents one record.

Example:
One student's details:


This entire row is one observation.

```python
import pandas as pd

students = pd.DataFrame({
    "Name":["A","B"],
    "Marks":[40,45]
})

print(students.iloc[0])
```

### Dataset
A dataset is a collection of observations organized in rows and columns.

Rows = Observations

Columns = Variables

Example:

```python
import pandas as pd

data = pd.DataFrame({
    "Name":["A","B","C"],
    "Age":[22,23,21]
})

print(data)
```

### Quantitative Data
Quantitative data consists of numerical values that can be measured or counted.

Types:

* Continuous
* Discrete

Example:

* Height

* Salary

```python
salary = [2000,3000,4000]

print(salary)
```

### Qualitative Data

Example:

Usually measured.

### Discrete Data
Discrete data contains whole numbers obtained by counting.

No decimal values.

Like:

Number of books

Example:

A classroom has 35 students.

```python
students = 35

print(students)
```
