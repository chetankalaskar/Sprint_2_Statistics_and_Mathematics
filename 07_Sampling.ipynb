# Sampling

### What is Sampling?
Sampling is the process of selecting a subset (sample) from a larger group (population) to collect information and draw conclusions about the entire population.

Sampling saves time, money, and effort compared to collecting data from everyone.

### Importance of Sampling

Sampling is important because it:

- Saves time.
- Reduces cost.
- Reduces effort.
- Makes large studies practical.
- Allows researchers to study large populations.
- Supports statistical analysis.
- Helps in decision-making.
- Is widely used in surveys and research.
- Is useful in Data Science and Machine Learning.


### Real-Life Applications of Sampling
### Education

Selecting students to study satisfaction with college facilities.

### Healthcare

Selecting patients to study satisfaction with hospital services.

### Business

Selecting customers to understand product satisfaction.

### Manufacturing

Inspecting a sample of products for quality control.

### Banking

Analyzing a sample of transactions for research or quality checks.

### Market Research

Selecting customers to understand preferences.

### Data Science

Selecting a manageable portion of a large dataset for analysis.



Example:

A college has 1,000 students, and the principal wants to conduct a survey about the quality of the college canteen. Since surveying all 1,000 students takes time, a sample of students is selected using different sampling techniques.

```python
import random

students = list(range(1, 1001))

sample = random.sample(students, 100)
print(sample[:10])   # Display first 10 selected students
```

A sample of 100 students is selected from the population of 1,000 students

### Random Sampling
Random Sampling is a sampling method in which every member of the population has an equal chance of being selected.

### Key Characteristics
- Every member has an equal chance of selection.
- Selection is based on randomization.
- Reduces selection bias.
- Simple to understand and implement.
- Requires a suitable sampling frame (list or source from which the sample is selected).



### Real-Life Applications
- Selecting students for a college survey.
- Selecting customers for feedback.
- Selecting products for quality inspection.
- Selecting households for a survey.

Example:

The principal randomly selects 100 students from the list of 1,000 students.

```python
import random

students = list(range(1, 1001))

random_sample = random.sample(students, 100)
print(random_sample[:10])
```

Every student has an equal probability of being selected.

### Stratified Sampling
Stratified Sampling divides the population into groups (strata) based on a common characteristic, and samples are taken from each group.


##3 Key Characteristics
* Population is divided into strata.
* Members within a stratum share a common characteristic.
* Samples are selected from each stratum.
* Helps ensure important groups are represented.
* Sampling can be equal or proportional.


### Real-Life Applications
- Selecting students from every academic year.
- Selecting employees from every department.
- Selecting customers from different age groups.
- Selecting samples from different income groups.


### Important Point

Stratified Sampling = Take samples from every important group.

Example:

The college has students from:

* First Year
* Second Year
* Third Year
* Fourth Year

The principal selects 25 students from each year, resulting in 100 students.

```python
first_year = list(range(1, 251))
second_year = list(range(251, 501))
third_year = list(range(501, 751))
fourth_year = list(range(751, 1001))

import random

sample = (
    random.sample(first_year, 25) +
    random.sample(second_year, 25) +
    random.sample(third_year, 25) +
    random.sample(fourth_year, 25)
)

print(sample[:10])
```

Every academic year is represented equally in the sample.

### Cluster Sampling
Cluster Sampling divides the population into clusters, then randomly selects one or more entire clusters for the study.

Population → All students

Clusters → Classrooms

Selected clusters → 2 classrooms

Sample → Students from selected classrooms


### Key Characteristics
- Population is divided into natural groups.
- Entire groups are selected randomly.
- Usually easier and cheaper than selecting individuals from the entire population.
- Useful when the population is geographically distributed.

### Real-Life Applications
- Selecting classrooms and surveying all students.
- Selecting villages and surveying households.
- Selecting company branches and surveying employees.
- Selecting hospitals and studying patients.


### Important Point

Cluster Sampling = Select whole groups.

Example:

The college has 20 classrooms.

The principal randomly selects 2 classrooms and surveys every student in those classrooms.

```python
import random

clusters = ["Class A", "Class B", "Class C", "Class D", "Class E"]

selected_cluster = random.choice(clusters)
print(selected_cluster)
```

Instead of selecting individual students, entire classrooms are selected

### Systematic Sampling
Systematic Sampling selects every kth member from the population after choosing a starting point.


### Key Characteristics
- Selects members at regular intervals.
- Uses a sampling interval called k.
- Usually begins with a starting point.
- Easy to implement.
- Useful when the population is arranged in a list.

Formula:
k = N / n

Where:

- N = Population size
- n = Required sample size
- k = Sampling interval

### Real-Life Applications
- Selecting every 10th customer.
- Inspecting every 20th product.
- Checking every 50th transaction.
- Selecting every 10th student from a list.


Example:

The principal selects every 10th student from the student register.

```python
students = list(range(1, 1001))

systematic_sample = students[::10]

print(systematic_sample[:10])
```

Students are selected at regular intervals.

### Sampling Bias
Sampling Bias occurs when some members of the population have a higher chance of being selected than others, making the sample unrepresentative.

### Common Types of Sampling Bias
### Selection Bias

Occurs when the selection process favors certain members.

Example: Selecting only students who are easy to contact.

### Undercoverage Bias

Occurs when some groups in the population are not included in the sampling frame.

Example: Conducting an online survey when some students do not have internet access.

### Non-Response Bias

Occurs when selected people do not respond and the non-responders differ from responders.

Example: People with negative opinions may be less likely to respond to a survey.

### Voluntary Response Bias

Occurs when people choose themselves to participate.

Example: Only students who strongly like or dislike the canteen voluntarily participate in a survey.

### How to Reduce Sampling Bias?
- Use appropriate random sampling.
- Use a complete sampling frame.
- Include important subgroups.
- Follow up with non-responders.
- Avoid selecting only easily accessible members.
- Use an appropriate sample size.


### What Happens When Sampling Bias Occurs?

The sample may give a systematically incorrect picture of the population.

Increasing the sample size does not necessarily remove sampling bias if the same biased selection method is used.

### Important Point

Bias = Problem with the way the sample is selected.

Example:

The principal surveys only final-year students about the canteen.

The opinions of first-, second-, and third-year students are ignored.

```python
final_year_students = list(range(751, 1001))
print(final_year_students[:10])
```

The sample does not represent the entire student population.

### Sampling Error
Sampling Error is the difference between the results obtained from a sample and the actual population value. 

It occurs because we study only a sample instead of the entire population.

### Key Characteristics
- Occurs because only a sample is studied.
- Can occur even when random sampling is used correctly.
- Changes from one sample to another.
- Usually decreases as sample size increases.
- Represents random variation rather than deliberate selection bias.


Uses:
- Measuring how close a sample estimate is to the population value.
- Understanding uncertainty in survey results.
- Comparing sampling methods.
- Evaluating statistical estimates.


### What Happens When It Goes Wrong?

A small sample may produce a result that is quite different from the population value.

Common Example

Suppose:

* Population average satisfaction = 8.5
* Sample average satisfaction = 8.1

The difference (0.4) is the sampling error.

```python
population_mean = 8.5
sample_mean = 8.1

sampling_error = population_mean - sample_mean

print(sampling_error)
```

A small sampling error indicates that the sample closely represents the population.

### Sample Size
Sample Size is the number of observations selected from the population for analysis.

### Why is Sample Size Important?
- A very small sample may produce unstable results.
- A larger sample generally provides more reliable estimates.
- Larger samples usually reduce sampling variability.
- Very large samples can increase cost and time.
- Appropriate sample size depends on the purpose of the study.


### What Happens When Sample Size is Too Small?
- Results may vary significantly from sample to sample.
- Estimates may be less precise.
- Important groups may not be adequately represented.


### What Happens When Sample Size is Too Large?
- More time may be required.
- More money may be required.
- Data collection and processing become more difficult.
- Resources may be wasted if a smaller sample would provide sufficient information.



### Important Point

- A larger sample does not automatically remove sampling bias.

- A large biased sample can still produce misleading results.

Example:

* Population = 1,000 students
* Sample Size = 100 students

```python
students = list(range(1, 1001))

sample = students[:100]

print("Sample Size:", len(sample))
```

The survey is conducted using 100 students out of the total 1,000 students.
