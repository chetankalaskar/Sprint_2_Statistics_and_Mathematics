### 1. What is Statistics?
Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data to make informed decisions.

### 2. Differentiate between Descriptive Statistics and Inferential Statistics.
Descriptive Statistics:

* Summarizes data
* Uses charts and summary measures

Example: Mean salary

Inferential Statistics:

* Draws conclusions about a population
* Uses hypothesis testing and probability

Example: Predict average salary of all employees

### 3. Differentiate between Population and Sample.
Population:

* Entire group
* Usually very large

Example: All employees in a company

Sample:

* Subset of the population
* Smaller and easier to analyze

Example: 50 employees selected for analysis


### 4. Explain the difference between Mean, Median, and Mode.
* Mean → Average of all values.
* Median → Middle value after sorting.
* Mode → Most frequently occurring value.

The mean is affected by outliers, while the median is more robust.

### 5. What is Variance?
Variance measures how far data values are spread from the mean. A higher variance indicates greater variability.

### 6. What is Standard Deviation?

Standard deviation is the square root of variance. It shows the average distance of data points from the mean.

### 7. What is the Interquartile Range (IQR)?

IQR is the difference between the third quartile (Q3) and the first quartile (Q1). It measures the spread of the middle 50% of the data and is commonly used to detect outliers.

### 8. What is Skewness?

Skewness measures the asymmetry of a distribution.

* Positive Skew → Long right tail
* Negative Skew → Long left tail
* Zero Skew → Symmetric distribution

### 9. Explain Bayes' Theorem.

Bayes' Theorem calculates the probability of an event based on prior knowledge.

Example:

A medical test estimates the probability that a person has a disease after receiving a positive test result.

### 10. What is Correlation?

Correlation measures the strength and direction of the relationship between two variables.

Correlation ranges from -1 to +1.

### 11. What is Multicollinearity?

Multicollinearity occurs when independent variables are highly correlated with each other, making it difficult to estimate the effect of each feature accurately.

### 12. What is Feature Importance?

Feature Importance indicates how much each feature contributes to predicting the target variable in a machine learning model.

### 13. What is a p-value?

The p-value is the probability of observing the obtained result if the null hypothesis is true.

* p-value < 0.05 → Reject the Null Hypothesis
* p-value ≥ 0.05 → Fail to Reject the Null Hypothes

### 14. Differentiate between Type I Error and Type II Error.
Type I Error:

* Reject a true Null Hypothesis
* False Positive

Type II Error:

* Fail to reject a false Null Hypothesis
* False Negative

### 15. When is ANOVA used?

ANOVA is used to compare the means of three or more groups to determine whether at least one group mean differs significantly from the others.

### 21. Detect Outliers using the IQR Method.

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower) | (df["Salary"] > upper)]

outliers
```

### 24. Perform a T-Test between HR and IT Salaries.

```python
from scipy.stats import ttest_ind

hr = df[df["Department"] == "HR"]["Salary"]
it = df[df["Department"] == "IT"]["Salary"]

t_stat, p_value = ttest_ind(hr, it)

print("T Statistic:", t_stat)
print("P Value:", p_value)
```
