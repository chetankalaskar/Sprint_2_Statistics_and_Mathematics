# Measures of Central Tendency

### What are Measures of Central Tendency?
Measures of Central Tendency are statistical measures that represent the center or typical value of a dataset. They summarize a large amount of data into a single representative value.


### Why are they important?
* Summarize large datasets.
* Compare different datasets.
* Identify the typical value.
* Used in Business Analytics, Finance, Healthcare, and AI/ML.

 ## The most common measures are:

* Mean
* Median
* Mode
* Weighted Mean
* Geometric Mean
* Harmonic Mean

### Mean (Arithmetic Mean)
The Mean is the average of all values in a dataset. It is calculated by adding all the observations and dividing by the total number of observations.

It is the most commonly used measure of central tendency.

Formula:

Where:

Σx = Sum of all observations

n = Number of observations

Example:

Marks of five students:
35, 40, 45, 80, 70

Sum = 35 + 40 + 45 + 80 + 70 = 270

Mean = 270 / 5 = 54

Mean = 54


```python
import numpy as np

marks = [35, 40, 45, 80, 70]

print(np.mean(marks))
```

Usage:

Before training a model, missing numerical values are often replaced with the mean of the column.

### Median
The Median is the middle value of a dataset after arranging the values in ascending order.

If there are:

Odd number of values → middle value
Even number of values → average of two middle values

Formula:

For odd observations

Middle value

For even observations


Odd Example:

Dataset:
12, 18, 20, 25, 30

Already sorted

Middle value = 20

Median = 20

Even example:

Dataset:
10, 20, 30, 40

Median:
(20 + 30)/2 = 25

```python
import numpy as np

numbers = [12, 18, 20, 25, 30]

print(np.median(numbers))
```

Usage:

Median is commonly used for replacing missing values when the data contains outliers.

### Mode
The Mode is the value that appears most frequently in a dataset.

A dataset may have:

One mode (Unimodal)
Two modes (Bimodal)
Multiple modes (Multimodal)
No mode

Formula:

No mathematical formula.

It is simply the value with the highest frequency.

Numerical Example

Dataset:
2, 4, 5, 5, 6, 7, 5

Frequency:
2, 4, 5, 5, 6, 7, 5

Mode = 5

```python
import pandas as pd

numbers = pd.Series([2,4,5,5,6,7,5])

print(numbers.mode())
```

Usage:

For categorical columns like Gender or City, missing values are often replaced using the mode.

### Weighted Mean
The Weighted Mean assigns different importance (weights) to different values.

Values with larger weights contribute more to the final average.

Formula:

Where:

x = value
w = weight

Example:

Calculation :

(90×4)+(80×3)+(70×2)=740

Total credits=9

Weighted Mean=740/9=82.22

```python
import numpy as np

marks = [90,80,70]
weights = [4,3,2]

print(np.average(marks, weights=weights))
```

Usage:

Weighted averages are used when combining predictions from multiple models (ensemble learning) or when giving more importance to certain observations.


### Geometric Mean
The Geometric Mean is the nth root of the product of all observations.

It is mainly used for growth rates, percentages, and financial returns.

Formula:

Example:

Numbers-2,4,8

Product = 2×4×8 = 64

Product = 2×4×8 = 64

Geometric Mean = 4

```python
!pip install scipy
```

```python
from scipy.stats import gmean

numbers = [2,4,8]

print(gmean(numbers))
```

Usage:

Used when evaluating multiplicative growth, probabilities, or ratios in machine learning and data analysis.

### Harmonic Mean
The Harmonic Mean is the reciprocal of the arithmetic mean of the reciprocals of the observations.

It is useful for rates such as speed, price per unit, or ratios.

Formula

Example :

Speeds= 40 km/h, 60 km/h

Calculation:

HM = 2 / (1/40 + 1/60)

HM = 48 km/h

```python
from scipy.stats import hmean

numbers = [40,60]

print(hmean(numbers))
```

Usage:

The F1-score, a widely used classification metric, is the harmonic mean of Precision and Recall.

For example, if:

Precision = 0.80
Recall = 0.60

Then:

The F1-score balances both metrics and is especially useful for imbalanced datasets.
