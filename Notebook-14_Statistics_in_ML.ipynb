# Statistics in Machine Learning

### Role of Statistics in Machine Learning
Statistics helps us collect, analyze, summarize, and interpret data. In Machine Learning, statistics is used to understand data, preprocess it, build models, and evaluate their performance.

### 1. Data Cleaning
Data cleaning is the process of identifying and correcting inaccurate, duplicate, or inconsistent data before training a machine learning model.

Example:

A customer dataset contains duplicate records and invalid ages such as -5 or 150. These values should be corrected or removed.

```python
import pandas as pd

data = {
    "Age": [25, 30, None, 30, 45],
    "Salary": [30000, 40000, 35000, 40000, 50000]
}

df = pd.DataFrame(data)

df = df.drop_duplicates()

print(df)
```

How Statistics Helps:

* Identifies incorrect values using statistical summaries.
* Detects duplicate and inconsistent records.
* Improves data quality before model training.

### 2. Missing Value Handling
Missing value handling is the process of replacing or removing missing values from a dataset.

Example:

Replace missing employee salaries with the mean salary.

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print (df["Salary"])
```

How Statistics Helps:

Statistics provides methods like mean, median, and mode to fill missing values.

### 3. Outlier Detection
Outlier detection identifies unusually high or low values that differ significantly from the rest of the data.

Example:

A salary of ₹10,00,000 in a dataset where most salaries are around ₹50,000 is an outlier.

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Salary"])
plt.show()
```

How Statistics Helps:

Statistical methods such as IQR, Z-score, and Standard Deviation help detect outliers.

### 4. Feature Engineering
Feature engineering is the process of creating new features from existing data to improve model performance.

Example:

Create a new feature:

Experience Level = Years of Experience × Skill Score

```python
df["Bonus"] = df["Salary"] * 0.10

print(df)
```

How Statistics Helps:

Statistical analysis helps identify which features are most useful for prediction.

### 5. Feature Scaling
Feature scaling adjusts feature values so they have a similar range.

Example:

Salary ranges from ₹20,000–₹80,000, while experience ranges from 1–10 years. Scaling prevents salary from dominating the model.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df[["Salary"]])
print(scaled_data)
```

How Statistics Helps:

Uses the mean and standard deviation to standardize data.

### 6. Data Normalization
Normalization rescales data to a fixed range, usually 0 to 1.

Example:

Convert salary values into the range 0–1.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

normalized = scaler.fit_transform(df[["Salary"]])
print(normalized)
```

How Statistics Helps:

Uses the minimum and maximum values to transform the data.

### 7. Model Evaluation
Model evaluation measures how well a machine learning model performs.

Example:

A model predicts employee salaries with 95% accuracy.

```python
from sklearn.metrics import accuracy_score

actual = [1,0,1,1]
predicted = [1,0,0,1]

print(accuracy_score(actual, predicted))
```

How Statistics Helps:

Statistical metrics such as Accuracy, Precision, Recall, F1-score, and RMSE evaluate model performance.

### 8. Linear Regression
Linear Regression predicts continuous numerical values using a straight-line relationship between variables.

Example:

Predict employee salary based on years of experience.

How Statistics Helps:

Uses correlation, variance, and least squares estimation to find the best-fit line.

### 9. Logistic Regression
Logistic Regression predicts categorical outcomes, such as Yes/No or 0/1.

Example:

Predict whether a customer will purchase a product.

How Statistics Helps:

Uses probability and the sigmoid function to estimate class probabilities.

### 10. Naive Bayes
Naive Bayes is a classification algorithm based on Bayes' Theorem.

Example:

Classify an email as Spam or Not Spam.

How Statistics Helps:

Uses conditional probability and assumes that features are statistically independent.

### 11. Clustering
Clustering groups similar data points without predefined labels.

Example:

Group customers based on their purchasing behavior.

How Statistics Helps:

Uses statistical measures such as distance, mean, and variance to form clusters.

### 12. Principal Component Analysis (PCA)
PCA reduces the number of features while preserving as much information as possible.

Example:

Reduce a dataset with 20 features to 5 principal components.

How Statistics Helps:

Uses variance, covariance, and eigenvalues to identify the most informative components.

### 13. Recommendation Systems
Recommendation systems suggest products, movies, or other items based on user preferences and behavior.

Example:

Netflix recommends movies based on your viewing history.

How Statistics Helps:

Statistics is used to calculate similarity, correlation, and user preferences to generate recommendations.
